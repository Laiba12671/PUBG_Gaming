# QUESTIONS:

### Q1: Who are the Top 10 Players with the Highest Kills?

```python
top_players = df.nlargest(10, "Kills")

plt.figure(figsize=(10,5))
sns.barplot(data=top_players, x="Player_ID", y="Kills")
plt.title("Top 10 Players with Highest Kills")
plt.show()
```

## This graph shows the top 10 players with the highest kills in PUBG.

### Q2: Which Map is Most Popular?

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x="Map")
plt.title("Most Popular PUBG Maps")
plt.show()
```

### This graph shows the number of players on each PUBG map.

### Q3: Which Tier Has the Highest Average Damage?

```python
plt.figure(figsize=(8,5))
sns.barplot(data=df, x="Tier", y="Damage_Dealt")
plt.title("Average Damage by Tier")
plt.show()
```

### This graph compares the average damage dealt by players in different tiers.

### Q4: What is the Relationship Between Kills and Damage?

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Kills", y="Damage_Dealt")
plt.title("Kills vs Damage Dealt")
plt.show()
```

### This graph shows the relationship between player kills and damage dealt.

### Q5: Which is the Most Used Favorite Weapon?

```python
plt.figure(figsize=(10,5))
sns.countplot(data=df, x="Favorite_Weapon")
plt.title("Most Popular Weapons")
plt.xticks(rotation=45)
plt.show()
```

### This graph shows which weapons are most preferred by PUBG players.

### Q6: Which Tier Has the Highest Survival Time?

```python
plt.figure(figsize=(8,5))
sns.barplot(data=df, x="Tier", y="Survival_Time_Minutes")
plt.title("Average Survival Time by Tier")
plt.show()
```

### This graph compares the average survival time of players across different tiers.

### Q7: What is the Relationship Between Kills and Headshot Percentage?

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Kills", y="Headshot_Percentage")
plt.title("Kills vs Headshot Percentage")
plt.show()
```

### This graph shows how player kills are related to headshot percentage.

### Q8: How are Match Ranks Distributed?

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df, x="Match_Rank", bins=15, kde=True)
plt.title("Distribution of Match Ranks")
plt.show()
```

### This graph shows the distribution of players' match ranks.
